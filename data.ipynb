# Bevölkerungsdaten Brandenburg

Quelle: https://www.statistik-berlin-brandenburg.de/bevoelkerung/demografie/bevoelkerungsstand 

```python
import pandas as pd
import numpy as np
```

```python
bev = pd.read_excel("bevoelkerungsstand-lange-reihe.xlsx", sheet_name="7 ", skiprows=5, skipfooter=3, usecols="B:AO", dtype=str)
bev
```

```python
bev = bev.dropna()
bev = bev.rename(columns={"Unnamed: 1": "ARS", "Unnamed: 2": "Name", "2022²": 2022, "2023²": 2023})
bev = bev.drop("Unnamed: 3", axis=1)
bev = bev.drop("2011.1", axis=1)
bev
```

## Trainingsset

```python
years = bev.columns[2:]
cutoff = int(len(years) * 0.8)
train_years = years[:cutoff]
test_years = years[cutoff:]
print("Train:", train_years[0], "-", train_years[-1])
print("Test:", test_years[0], "-", test_years[-1])
```

```python
window_size = 5

def create_windows(data, years):
    values = data[years].to_numpy()

    rows = []

    for row in values:
        for i in range(len(row) - window_size):
            rows.append(
                [*row[i:i + window_size], row[i + window_size]]
            )

    return pd.DataFrame(
        rows,
        columns=[f"X{i}" for i in range(window_size)] + ["y"]
    )
```

```python
train = create_windows(bev, train_years)
test = create_windows(bev, test_years)
```

```python
train = train.apply(pd.to_numeric, errors="coerce")
train = train.dropna()
train
```

```python
test = test.apply(pd.to_numeric, errors="coerce")
test = test.dropna()
test
```

```python
train.to_csv("trainingsset.csv", index=False)
test.to_csv("testingset.csv", index=False)
```
