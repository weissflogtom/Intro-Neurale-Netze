# Bevölkerungsdaten Brandenburg

Quelle: https://www.statistik-berlin-brandenburg.de/bevoelkerung/demografie/bevoelkerungsstand 

```python
import pandas as pd
```

```python
bev = pd.read_excel("bevoelkerungsstand-lange-reihe.xlsx", sheet_name="7 ", skiprows=5, skipfooter=3, usecols="B:AO", dtype=str)
bev
```

```python
bev = bev.dropna()
bev = bev.rename(columns={"Unnamed: 1": "ARS", "Unnamed: 2": "Name", "2022²": "2022", "2023²": "2023"})
bev = bev.drop("Unnamed: 3", axis=1)
bev
```

## Trainingsset

```python
window_size = 5

values = bev.iloc[:, 2:].to_numpy()

trainingsset = pd.DataFrame(
    [
        [*row[i:i + window_size], row[i + window_size]]
        for row in values
        for i in range(len(row) - window_size)
    ],
    columns=[f"X{i}" for i in range(window_size)] + ["y"]
)

trainingsset
```

```python
trainingsset = trainingsset.apply(pd.to_numeric, errors="coerce")
trainingsset = trainingsset.dropna()
trainingsset
```

```python
trainingsset.to_csv("trainingsset.csv", index=False)
```
